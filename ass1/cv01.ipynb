### **OpenCV**

```python
image_path = r'images/Lenna.png'
```

```python
img = cv2.imread(image_path)
```

```python
img.shape 
```

```python
plt.figure(figsize=(15, 6))

plt.imshow(img, cmap='grey')
plt.title("Original Image")
plt.axis('off')
```

```python
H, W, C = img.shape
```

```python
# 1. Image Slicing (Region of Interest - ROI)
# Select a central 50x50 area of the original image
# Note: H, W were defined in Section 2
start_row = H // 4
end_row = 3 * H // 4
start_col = W // 4
end_col = 3 * W // 4

roi = img[start_row:end_row, start_col:end_col] # Slicing for rows and columns
print(f"ROI Shape: {roi.shape}")
print("-" * 30)
```

```python

R_channel = img[:, :, 0]

red_only_img = img.copy()
red_only_img[:, :, 1] = 0 # Green channel set to zero
red_only_img[:, :, 2] = 0 # Blue channel set to zero
print("Red Channel Max Value:", np.max(R_channel))
```

```python
# 3. Simple Mathematical Operation (Brightness Increase)
# Add a constant value (e.g., 50) to all pixel intensities
# np.clip is used to keep values within the valid range [0, 255]
brightness_increase = np.clip(img.astype(np.int16) + 50, 0, 255).astype(np.uint8)
print("Brightness Increased. Check the displayed image.")
```

```python

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(roi)
plt.title("ROI (Slicing)")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(red_only_img)
plt.title("Channel Manipulation (Red Only)")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(brightness_increase)
plt.title("Mathematical Op. (Brightness +50)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## 1. Read the image

```python
img = cv2.imread('images/Lenna.png')
```

## Convert from BGR (OpenCV default) to RGB for proper display

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

## 2. Approximate region of the eyes (manually chosen based on Lenna's image)
## Coordinates are in [y1:y2, x1:x2] format

```python
y1, y2 = 250, 290
x1, x2 = 240, 355
```

## Extract the eye region

```python
eye_region = img_rgb[y1:y2, x1:x2]
```

## 3. Reduce brightness in that region (divide intensity by 10)

```python
img_modified = img_rgb.copy()
img_modified[y1:y2, x1:x2] = (eye_region / 10).astype(np.uint8)
```

## 4. Display the original and modified images side by side

```python
plt.figure(figsize=(12, 6))
```

## Original image

```python
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
```

## Modified image with darker eyes

```python
plt.subplot(1, 2, 2)
plt.imshow(img_modified)
plt.title("Brightness Reduced in Eye Region (×0.1)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

<h2>ساخت تصویر رنگی از سه کانال</h2>
</div>



```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Create three 50×50 single-channel grayscale arrays
## R_arr filled with 255 (pure red)
## G_arr filled with 100 (weak green)
## B_arr filled with 0 (no blue)

```python
R_arr = np.full((50, 50), 255, dtype=np.uint8)
G_arr = np.full((50, 50), 100, dtype=np.uint8)
B_arr = np.zeros((50, 50), dtype=np.uint8)
```

## 2. Resize each array to 100×100 using tiling

```python
R_arr = np.tile(R_arr, (2, 2))
G_arr = np.tile(G_arr, (2, 2))
B_arr = np.tile(B_arr, (2, 2))
```

## 3. Merge the three single channels into one RGB image


```python
color_img = np.stack((R_arr, G_arr, B_arr), axis=-1)
```

## 4. Display the final color image

```python
plt.imshow(color_img)
plt.title("Combined RGB Image")
plt.axis('off')
plt.show()
```
